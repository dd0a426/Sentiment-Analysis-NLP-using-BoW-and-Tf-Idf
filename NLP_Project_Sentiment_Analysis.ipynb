```python
 df['text']=df['text'].apply(lambda x: x.lower())
```

Removing Punctiation

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df["text"]=df["text"].apply(remove_punc)
```

Removing numbers

```python
def remove_numbers(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new
df["text"]=df["text"].apply(remove_numbers)
```

Remove Emoji's

```python
def remove_emojis(txt):
  new=""
  for i in txt:
    if i.isascii():
      new=new+i
  return new

df["text"]=df["text"].apply(remove_emojis)
```

```python
#now remove stop words from text
def remove(txt):
  #words=word_tokenize(txt)
  words=txt.split()
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return " ".join(cleaned)
```

Applying User Input for sentiment Analysis

```python
import re
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Ensure NLTK resources are downloaded
import nltk
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    text = text.lower() # Convert to lowercase
    text = text.translate(str.maketrans('', '', string.punctuation)) # Remove punctuation
    text = re.sub(r'\d+', '', text) # Remove numbers
    text = ''.join(char for char in text if char.isascii()) # Remove non-ascii (emojis)
    words = text.split() # Split into words (similar to how stopwords were removed)
    cleaned_words = [word for word in words if word not in stop_words] # Remove stopwords
    return ' '.join(cleaned_words)
```

```python
def predict_emotion(user_text):
    preprocessed_text = preprocess_text(user_text)
    # Transform the text using the trained TF-IDF vectorizer
    user_text_tfidf = tfidf_vectorizer.transform([preprocessed_text])

    # Predict the emotion using the trained Logistic Regression model
    prediction_numeric = logistic_model.predict(user_text_tfidf)

    # Map the numeric prediction back to the original emotion label
    # Invert the emotion_numbers dictionary to map numbers back to emotions
    emotion_labels = {v: k for k, v in emotion_numbers.items()}
    predicted_emotion = emotion_labels[prediction_numeric[0]]

    return predicted_emotion
```

```python
from ipywidgets import Text, Output
from IPython.display import display

text_input = Text(description='Enter text:')
output_area = Output()

def on_text_change(change):
    with output_area:
        output_area.clear_output()
        user_input = change['new']
        if user_input:
            sentiment = predict_emotion(user_input)
            print(f'Your text: "{user_input}"')
            print(f'Predicted emotion: {sentiment}')

text_input.observe(on_text_change, names='value')
display(text_input, output_area)
```

## Save and Load the Trained Model

```python
import joblib

# Define filenames for saving the model and vectorizer
model_filename = 'logistic_model_tfidf.joblib'
vectorizer_filename = 'tfidf_vectorizer.joblib'

# Save the trained logistic regression model (with TF-IDF)
joblib.dump(logistic_model, model_filename)
print(f'Logistic Regression model saved as {model_filename}')

# Save the TF-IDF vectorizer
joblib.dump(tfidf_vectorizer, vectorizer_filename)
print(f'TF-IDF vectorizer saved as {vectorizer_filename}')
```

```python
# Load the model and vectorizer
loaded_model = joblib.load(model_filename)
loaded_vectorizer = joblib.load(vectorizer_filename)

print(f'Model loaded: {loaded_model}')
print(f'Vectorizer loaded: {loaded_vectorizer}')
```
